# Première Etape

```python
import nltk
```

```python
text = """" The first thing that you need to do is to get yourself a good computer. There are many different types of computers that you can choose from, so it is important that you find one that is right for you. You will need to make sure that the computer has enough memory and processing power to handle the tasks that you will be doing. You will also need to make sure that the computer has a good graphics card so that you can play games and watch videos. Once you have found a computer that is right for you, you will need to set it up. This will involve installing the operating system, setting up the internet connection, and installing any software that you will need. Once you have done this, you will be ready to start using your computer. """
```

```python
nltk.download('punkt')
```

```python
nltk.download('punkt_tab')
```

```python
nltk.sent_tokenize(text)
textTokernizer = nltk.word_tokenize(text)
print(textTokernizer)
```
